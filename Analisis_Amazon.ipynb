```python
import pandas as pd

with zipfile.ZipFile('archive.zip') as z:
    z.extractall('data')

df = pd.read_csv('data/Amazon.csv')
print('Baris dan kolom:', df.shape)
print(list(df.columns))
df.head()
```

```python
print(df.dtypes)
print('Duplikat baris persis:', df.duplicated().sum())
print('OrderID unik:', df['OrderID'].nunique(), 'dari', len(df), 'baris')
print(df.isna().sum().sort_values(ascending=False))
```

```python
print('Macam status:', df['OrderStatus'].nunique())
print(df['OrderStatus'].value_counts())
print('Kategori:', df['Category'].nunique(), '| Negara:', df['Country'].nunique(), '| Metode bayar:', df['PaymentMethod'].nunique())
tgl = pd.to_datetime(df['OrderDate'], errors='coerce')
print('Tanggal gagal dibaca:', tgl.isna().sum())
print('Rentang tanggal:', tgl.min(), 'sampai', tgl.max())
```

```python
hitung = df['Quantity'] * df['UnitPrice'] * (1 - df['Discount']) + df['Tax'] + df['ShippingCost']
selisih = (hitung - df['TotalAmount']).abs()
print('Cocok (selisih di bawah 0.01):', (selisih < 0.01).sum())
print('Meleset:', (selisih >= 0.01).sum())
print('Selisih terbesar:', selisih.max())
```

```python
print(df[['Quantity','UnitPrice','Discount','Tax','ShippingCost','TotalAmount']].describe().loc[['min','max']])
print('Diskon di luar 0 sampai 1:', ((df['Discount'] < 0) | (df['Discount'] > 1)).sum())
print('Harga atau jumlah tidak positif:', ((df['UnitPrice'] <= 0) | (df['Quantity'] <= 0)).sum())
```

```python
per_produk = df.groupby('ProductName')['Category'].nunique()
print('Produk yang kategorinya lebih dari 1:', (per_produk > 1).sum(), 'dari', len(per_produk), 'nama produk')
print(per_produk[per_produk > 1])

per_id = df.groupby('ProductID')['ProductName'].nunique()
print('ProductID yang namanya lebih dari 1:', (per_id > 1).sum(), 'dari', len(per_id), 'ID')
```

```python
per_kota = df.groupby('City')['Country'].nunique()
print('Kota yang negaranya lebih dari 1:', (per_kota > 1).sum(), 'dari', len(per_kota), 'kota')
per_state = df.groupby('City')['State'].nunique()
print('Kota yang statenya lebih dari 1:', (per_state > 1).sum())
per_brand = df.groupby('ProductName')['Brand'].nunique()
print('Produk yang brandnya lebih dari 1:', (per_brand > 1).sum(), 'dari', len(per_brand), 'produk')
```

```python
per_cust = df.groupby('CustomerID')['CustomerName'].nunique()
print('CustomerID yang namanya lebih dari 1:', (per_cust > 1).sum(), 'dari', len(per_cust), 'pelanggan')
harga = df.groupby('ProductName')['UnitPrice'].agg(['min', 'max', 'nunique'])
print(harga)
```

```python
kerja = df.copy()
kerja['Tanggal'] = pd.to_datetime(kerja['OrderDate'])
kerja['Bulan'] = kerja['Tanggal'].dt.strftime('%Y-%m')
print('Baris kerja:', len(kerja))
print('Bulan kosong:', kerja['Bulan'].isna().sum())
```

```python
selesai = kerja[kerja['OrderStatus'] == 'Delivered']
tren = pd.DataFrame({
    'Pesanan': kerja.groupby('Bulan')['OrderID'].count(),
    'Pendapatan': selesai.groupby('Bulan')['TotalAmount'].sum(),
})
print(tren)
print('Pendapatan tertinggi:', tren['Pendapatan'].idxmax(), tren['Pendapatan'].max())
print('Pendapatan terendah:', tren['Pendapatan'].idxmin(), tren['Pendapatan'].min())
```

```python
kerja['Tahun'] = kerja['Tanggal'].dt.year
jumlah = pd.crosstab(kerja['Tahun'], kerja['OrderStatus'])
persen = pd.crosstab(kerja['Tahun'], kerja['OrderStatus'], normalize='index') * 100
print(jumlah)
print(persen.round(2))
```

```python
prod = kerja.groupby('ProductName').agg(Pesanan=('OrderID', 'count'))
prod['Pendapatan'] = selesai.groupby('ProductName')['TotalAmount'].sum()
print('Menurut jumlah pesanan:')
print(prod.sort_values('Pesanan', ascending=False).head(10))
print('Menurut pendapatan:')
print(prod.sort_values('Pendapatan', ascending=False).head(10))
```

```python
bayar = kerja.groupby('PaymentMethod').agg(Pesanan=('OrderID', 'count'))
bayar['Pendapatan'] = selesai.groupby('PaymentMethod')['TotalAmount'].sum()
bayar['Bagian'] = (bayar['Pesanan'] / bayar['Pesanan'].sum() * 100).round(2)
print(bayar.sort_values('Pesanan', ascending=False))
```

```python
s2 = kerja[kerja['OrderStatus'] == 'Delivered']
print('Total pendapatan Delivered:', s2['TotalAmount'].sum())
print('AOV Delivered keseluruhan:', round(s2['TotalAmount'].mean(), 2))
print('AOV per tahun:')
print(s2.groupby('Tahun')['TotalAmount'].mean().round(2))
print('Rata rata TotalAmount per status:')
print(kerja.groupby('OrderStatus')['TotalAmount'].mean().round(2))
```

```python
kerja.to_csv('amazon_kerja_bersih.csv', index=False)
tren.to_csv('amazon_tren_bulanan.csv')
prod.to_csv('amazon_produk.csv')
bayar.to_csv('amazon_metode_bayar.csv')
print('tersimpan')
```
